# restore

We cannot simply RESTORE the Bronze table because the table's history and the streaming checkpoints are separate systems. Auto Loader won't re-ingest the landing files because it thinks they are already processed, and the downstream Silver stream will ignore the rollback for the same reason

## Create a safe restore demo table

```python
%sql
CREATE OR REPLACE TABLE
ride_lakehouse_dev.bronze.drivers_restore_demo

USING DELTA

AS

SELECT *
FROM ride_lakehouse_dev.bronze.drivers;
```

First and only version of copied tavle equals to last version of our original table

```python
%sql
DESCRIBE HISTORY
ride_lakehouse_dev.bronze.drivers_restore_demo;
```

```python
demo_ids = (
    spark.table("ride_lakehouse_dev.bronze.drivers_restore_demo")
        .select("id","rating")
        .limit(2)
        .collect()
)

first_row = demo_ids[0]
second_row = demo_ids[1]

print(first_row["id"], first_row["rating"])
print(second_row["id"], second_row["rating"])
```

result:
D-20260929114913448298-000 2.0
D-20260929114913448298-001 1.34

```python
spark.sql(f"""
    UPDATE ride_lakehouse_dev.bronze.drivers_restore_demo

    SET rating = 5.0

    WHERE id = '{first_row["id"]}'
""")
```

```python
spark.sql(f"""
    SELECT id, rating FROM ride_lakehouse_dev.bronze.drivers_restore_demo
    WHERE id = '{first_row["id"]}'
""").show()
```

now we have two demo

```python
%sql
DESCRIBE HISTORY
ride_lakehouse_dev.bronze.drivers_restore_demo;
```

## restoring

```python
%sql
RESTORE TABLE
ride_lakehouse_dev.bronze.drivers_restore_demo

TO VERSION AS OF 0;
```

```python
spark.sql(f"""
    SELECT id, rating FROM ride_lakehouse_dev.bronze.drivers_restore_demo
    WHERE id = '{first_row["id"]}'
""").show()
```

as we can see, prvious version of rating displayed without time travel
