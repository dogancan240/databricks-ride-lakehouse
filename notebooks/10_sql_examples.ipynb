# DDL
I will create a table and alter it

````python
%sql
CREATE TABLE
ride_lakehouse_dev.gold.sql_practice
(
    review_id BIGINT,
    ride_id STRING,
    issue STRING
)
USING DELTA;
````

````python
%sql
DESCRIBE TABLE
ride_lakehouse_dev.gold.sql_practice;
````

Alter: Updating data schema not data

````python
%sql
ALTER TABLE
ride_lakehouse_dev.gold.sql_practice

ADD COLUMNS (
    review_status STRING
);
````

````python
%sql
DESCRIBE TABLE
ride_lakehouse_dev.gold.sql_practice;
````

# DQL


````python
%sql
SELECT
    driver_id,
    driver_name,
    total_rides,
    total_distance_km,
    total_revenue,
    average_ride_revenue
FROM ride_lakehouse_dev.gold.driver_stats
ORDER BY total_revenue DESC
LIMIT 10;
````

````python
%sql
SELECT
    ride_date,
    total_rides,
    unique_drivers,
    unique_customers,
    total_distance_km,
    total_revenue,
    average_fare,

    ROUND(
        total_revenue / NULLIF(total_rides, 0),
        2
    ) AS revenue_per_ride

FROM ride_lakehouse_dev.gold.daily_stats

ORDER BY ride_date DESC;
````

````python
%sql
DESCRIBE HISTORY
ride_lakehouse_dev.gold.sql_practice;
````
