This code doesn't consist only permition of SELECT. Databricks uses USE CATALOG, USE SCHEMA then SELECT. So it checks permitons for catalog and shcema.

# Grant USE to Catalog
with this code our user have access to whole catalog thanks to the its own group permitionsbut he still needs permiton for schema

````python
%sql
GRANT USE CATALOG
ON CATALOG ride_lakehouse_dev
TO `innowise`;
````

# Grand use to schema
We will give permition to schema but he still needs permition fro select

````python
%sql
GRANT USE SCHEMA
ON SCHEMA ride_lakehouse_dev.gold
TO `innowise`;
````
