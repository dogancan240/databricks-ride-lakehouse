# Creating catalog
A Catalog is the highest-level container in Unity Catalog. (e.g., `prod_data`, `dev_data`). But Our project is small so one catalog is enough. Catalog is used to separate enviroments. 

````python
%sql
CREATE CATALOG IF NOT EXISTS ride_lakehouse_dev
COMMENT 'Development environment for ride lakehouse practice';
````

````python
%sql
USE CATALOG ride_lakehouse_dev;
````

# Creating Schemas
```text
ride_lakehouse_dev
│
├── bronze
├── silver
└── gold
```

````python
%sql
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Raw and minimally processed data';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Cleaned, validated and deduplicated data';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Business-level aggregated data';
````

# Creating Volume
We use volume to handle incoming non-tabular data. 
```text
TABLE
    ↓
structured data

VOLUME
    ↓
files

Faker
   ↓
JSON
   ↓
Volume
   ↓
Spark
   ↓
Delta Table
```
drivers_001.json
drivers_002.json
rides_001.json
payments_001.json

````python
%sql
CREATE VOLUME IF NOT EXISTS
ride_lakehouse_dev.bronze.streaming_files
COMMENT 'Landing files, streaming checkpoints and Auto Loader schemas';
````

# Create folders inside the Volume
```text
landing/
   ↓
actual incoming data

checkpoints/
   ↓
what has already been processed?

schemas/
   ↓
what does incoming data look like?
```


````python
BASE_PATH = "/Volumes/ride_lakehouse_dev/bronze/streaming_files"

folders = [
    "landing/drivers",
    "landing/rides",
    "landing/payments",

    "checkpoints/bronze_drivers",
    "checkpoints/bronze_rides",
    "checkpoints/bronze_payments",

    "schemas/drivers",
    "schemas/rides",
    "schemas/payments",
]

for folder in folders:
    dbutils.fs.mkdirs(f"{BASE_PATH}/{folder}")
````
